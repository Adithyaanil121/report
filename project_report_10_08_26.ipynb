```python
from sklearn import datasets
import pandas as pd
```

```python
a=datasets.load_iris()

```

```python
x=a.data
y=a.target
```

```python
x=a.data[:,:2]
y=(a.target==0).astype(int)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
from sklearn.svm import SVC
```

```python
svm=SVC(kernel='linear',C=1)
```

```python
svm.fit(x_train,y_train)
```

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import accuracy_score
```

```python
y_prediction=svm.predict(x_test)
accuracy=accuracy_score(y_test,y_prediction)
print(f"accuracy:{accuracy}")
```

```python
plt.scatter(x[:,0],x[:,1],c=y)
w= svm.coef_[0]
b=svm.intercept_[0]
x_values=np.linspace(x[:,0].min(),x[:,0].max(),100)
y_decision=-(w[0]/w[1])*x_values-b/w[1]
y_margin1=-(w[0]/w[1])*x_values-(b-1)/w[1]
y_margin2=-(w[0]/w[1])*x_values-(b+1)/w[1]
plt.plot(x_values,y_decision,'k-')
plt.plot(x_values,y_margin1,'k--')
plt.plot(x_values,y_margin2,'k--')
plt.scatter(
    svm.support_vectors_[:,0],
    svm.support_vectors_[:,1],
    s=100,
    facecolors='none',
    edgecolors='red'
)
plt.xlabel("sepal length")
plt.ylabel("sepal width")
plt.title('linear svm decision boundary')
plt.show()
```
