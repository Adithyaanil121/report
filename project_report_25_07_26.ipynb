```python
x=data.drop('quality',axis=1)
y=(data['quality']>=6).astype(int)
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
scalar1=StandardScaler()
x_trainscaled=scalar1.fit_transform(x_train)
x_testscaled=scalar1.transform(x_test)
```

```python
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score
architecture=[
    (5,),
    (10,),
    (20,),
    (10,10),
    (20,10)
]
for arch in architecture:
  mlp=MLPClassifier(hidden_layer_sizes=arch,max_iter=1000,random_state=42)
  mlp.fit(x_trainscaled,y_train)
  y_pred=mlp.predict(x_testscaled)
  accuracy=accuracy_score(y_test,y_pred)
  print(f"Accuracy for architecture:",arch,"accuracy:",round(accuracy,4))
```
