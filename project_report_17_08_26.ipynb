```python
from sklearn import datasets
import pandas as pd
```

```python
iris=datasets.load_iris()
```

```python
x=iris.data
y=iris.target
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.transform(x_test)
```

```python
from sklearn.svm import SVC
```

```python
svm1=SVC(kernel='linear',C=1)
svm2=SVC(kernel='poly',degree=3)
svm3=SVC(kernel='rbf')
```

```python
model1=svm1.fit(x_train,y_train)
```

```python
model2=svm2.fit(x_train,y_train)
```

```python
model3=svm3.fit(x_train,y_train)
```

```python
from sklearn.metrics import accuracy_score
```

```python
prediction1= model1.predict(x_test)
accuracy1=accuracy_score(y_test,prediction1)
print(f"the accuracy of model 1 is {accuracy1}")
```

```python
prediction2=model2.predict(x_test)
accuracy2=accuracy_score(y_test,prediction2)
print(f"the accuracy of model2 is {accuracy2}")
```

```python
prediction3=model3.predict(x_test)
accuracy3=accuracy_score(y_test,prediction3)
print(f"the accuracy of model3 is {accuracy3}")
```

```python
print(f"the accuracy of linear is {accuracy1}")
print(f"the accuracy of polynomial is {accuracy2}")
print(f"the accuracy of rbf is {accuracy3}")
```
